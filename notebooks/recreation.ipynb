```python
import sys
from pathlib import Path
import pandas as pd

# Add repo root so we can import the authors' metric functions
ROOT = Path("..").resolve()
sys.path.append(str(ROOT))

from module.metrics import ARR, SR, MDD

# Load the authors' DBB price data
p = pd.read_parquet(
    "../datasets/processd_day_future_etfs/price/DBB.parquet"
)

# Keep the test period from 2021 onwards
test = p[p["timestamp"] >= "2021-01-01"].copy()

# Calculate daily DBB returns
rets = test["adj_close"].pct_change().dropna().to_numpy()

# Use the authors' exact metric functions
arr = ARR(rets)
sr = SR(rets)
mdd = MDD(rets)

# Results
print("Test period:", test["timestamp"].min(), "to", test["timestamp"].max())
print("Number of trading days:", len(test))
print()
print("ARR:", arr * 100, "%")
print("Sharpe:", sr)
print("MDD:", mdd * 100, "%")
```
